```python
# Get the article
from newspaper import Article

article = Article('https://www.mayoclinic.org/diseases-conditions/chronic-kidney-disease/symptoms-causes/syc-20354521')

article.download()
article.parse()
article.nlp()

corpus = article.text
```

```python
# Print the articles text
print(corpus)
```

```python
# Tokenization
from nltk.tokenize import sent_tokenize

sentence_list = sent_tokenize(corpus)
```

```python
# Print the list of sentences
print(sentence_list)
```

```python
# Function to return a random greeting response to usets greeting
def greeting_response(text):
  text = text.lower()

  # Bots greetings
  bot_greetings = ['howdy', 'hi', 'hey', 'hello', 'hola']
  # Users greetings
  user_greetings = ['hi', 'hey', 'hello', 'hola', 'greetings', 'wassup']


  for word in text.split():
    if word in user_greetings:
      return random.choice(bot_greetings)
```

```python
def index_sort(list_var):
  length = len(list_var)
  list_index = list(range(0, length))

  x = list_var
  for i in range(length):
    for j in range(length):
      if x[list_index[i]] > x[list_index[j]]:
        # Swap
        temp = list_index[i]
        list_index[i] = list_index[j]
        list_index[j] = temp


  return list_index
```

```python
# Create a bots response
def bot_response(user_input):

    user_input = user_input.lower()
    sentence_list.append(user_input)

    cm = CountVectorizer().fit_transform(sentence_list)
    similarity_scores = cosine_similarity(cm[-1], cm).flatten()

    # ignore user input itself
    similarity_scores[-1] = 0

    # also ignore very short / useless sentences
    filtered_scores = similarity_scores.copy()
    for i in range(len(sentence_list)):
        if len(sentence_list[i].split()) < 5:
            filtered_scores[i] = 0

    index = filtered_scores.argmax()
    score = filtered_scores[index]

    sentence_list.pop()

    if score < 0.15:
        return "I apologize, I do not understand."

    return sentence_list[index]
```

```python
# Start the chat
print('Doc Bot: I am Doctor Bot or Doc Bot for short. I will answer your queries about Chronic Kidney Disease. If you want to exit, type bye.')

exit_list = ['exit', 'see you later', 'quit', 'bye', 'break']

while(True):
  user_input = input()
  if user_input.lower() in exit_list:
    print('Doc Bot: Chat with you later !')
    break
  else:
    if greeting_response(user_input) != None:
      print('Doc Bot: '+greeting_response(user_input))
    else:
      print('Doc Bot: '+bot_response(user_input))
```
